In order to begin the process of characterizing the various accommodations available on Airbnb in the city of Amsterdam, the following hypotheses are of interest:  

1. 53% of the review scores for listings on the Airbnb platform fall between 20 and 96 points. 

2. The average rental price for a property that is above the 78th percentile of city prices is 300 euros. 

3. 20% of properties in the city are being rented for less than the Amsterdam city average. 

4. Alberto claims to have rented a property on the Airbnb platform for 130 euros, which is one standard deviation above the average. 

# 1. Import libraries

#### 2. The average rental price for a property that is above the 78th percentile of city prices is 300 euros.

```python
price_78 = data['price'].quantile(0.78)
mean_above_78 = data.loc[data['price'] > price_78, 'price'].mean()
print(f"Price at 78th percentile: {price_78:.2f}")
print(f"Average price above 78th percentile: {mean_above_78:.2f}")
```

```python
fig = px.histogram(data, x='price', nbins=500,
                   title='Distribution of Rental Prices',
                   labels={'price': 'Price (euros)'},
                   color_discrete_sequence=['steelblue'])

fig.add_vline(x=price_78, line_dash='dash', line_color='red', 
              annotation_text=f'78% ({price_78:.0f})')
fig.add_vline(x=average_above_78, line_dash='dash', line_color='orange', 
              annotation_text=f'Mean above 78% ({mean_above_78:.0f})',
             annotation_position='bottom right')

fig.show()
```

#### The hypothesis is correct: the average price for over 78% of the data points is 300.

#### 3. 20% of properties in the city are being rented for less than the Amsterdam city average.

```python
mean_price = data['price'].mean()
below_mean = data.loc[data['price'] < mean_price]
percentage = len(below_mean) / len(data) * 100

print(f"The mean is: {mean_price:.2f}")
print(f"Count below mean: {len(below_mean)}")
print(f"The percentage of rental income below the average is: {percentage:.2f}%")
```

```python
fig = px.histogram(data, x='price', nbins=500,
                   title=f'Distribution of Rental Prices - {percentage:.2f}% below mean',
                   labels={'price': 'Price (euros)'},
                   color_discrete_sequence=['steelblue'])

fig.add_vline(x=mean_price, line_dash='dash', line_color='red',
              annotation_text=f'Mean ({mean_price:.2f})',
              annotation_position='top right')

fig.show()
```

#### The statement is NOT TRUE. The actual percentage is 66.13%, which means the distribution is skewed to the right.

#### 4. Alberto claims to have rented a property on the Airbnb platform for 130 euros, which is one standard deviation above the average.

```python
std_price = data['price'].std()

one_std_above = mean_price + std_price

print(f"Mean + 1 std: {one_std_above:.2f}")
```

```python
fig, ax = plt.subplots(figsize=(12, 3))

data['price'].hist(bins=200, ax=ax, color='steelblue', edgecolor='white')

ax.axvline(x=mean_price, color='blue', linestyle='--', label=f'Mean ({mean_price:.2f})')
ax.axvline(x=one_std_above, color='green', linestyle='--', label=f'Mean + 1 std ({one_std_above:.2f})')
ax.axvline(x=130, color='red', linestyle='--', label="Alberto's price (130)")

ax.set_title("Alberto's Claim: 130 euros is one standard deviation above the mean")
ax.set_xlabel('Price (euros)')
ax.set_ylabel('Count')
ax.legend(loc='upper right')
plt.show()
```
