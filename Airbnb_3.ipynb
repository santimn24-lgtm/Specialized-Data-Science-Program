# Exercise 3

To begin characterizing the various types of accommodations available on Airbnb in Amsterdam, the following hypotheses are of interest:

1.    Review scores and communication scores are directly proportional.

2.    The relationship between rental price and the number of rooms is inversely proportional.

# 1. Import libraries

```python
import pandas as pd
import numpy as np
import plotly.express as px
```

# 2. Load data

```python
data= pd.read_csv('listings_m-2-.csv', sep=',')
data.head()
```

# 3. Hypothesis

#### 1. Review scores and communication scores are directly proportional.

```python
review_scores = data[['review_scores_rating', 'review_scores_communication', 'room_type']]

review_scores = review_scores[
    (review_scores['review_scores_rating'].notna()) &
    (review_scores['review_scores_communication'].notna()) &
    (review_scores['review_scores_rating'] != 0) &
    (review_scores['review_scores_communication'] != 0)
]
review_scores.head()
```

```python
fig = px.scatter(
    review_scores,
    x='review_scores_rating',
    y='review_scores_communication',
    color='room_type',
    trendline='ols',
    title='Relationship between Rating and Communication by Room Type'
)

fig.show()
```

```python
pearson_corr = review_scores['review_scores_rating'].corr(
    review_scores['review_scores_communication'],
    method='pearson'
)
print(f"Pearson correlation: {pearson_corr}")
```

```python
spearman_corr = review_scores['review_scores_rating'].corr(
    review_scores['review_scores_communication'],
    method='spearman'
)
print(f"Spearman correlation: {spearman_corr}")
```

#### When plotting the scatter plot between the variables “review scores rating” and “review scores communication,” a trend line with a positive slope is observed, indicating a direct relationship between the two variables. However, the calculation of the Pearson (0.658) and Spearman (0.433) correlation coefficients shows that this relationship is of moderate strength and not strong enough to confirm a direct proportionality. Therefore, the proposed hypothesis is rejected.

#### 2. The relationship between rental price and the number of rooms is inversely proportional.

```python
relationship = data[['price', 'bedrooms']]

relationship = relationship[
    (relationship['price'].notna()) &
    (relationship['bedrooms'].notna()) &
    (relationship['price'] != 0) &
    (relationship['bedrooms'] != 0) &
    (relationship['price'] <= 400) &
    (relationship['bedrooms'] <= 5)
]

relationship.head()
```

```python
fig = px.scatter(
    relationship,
    x='price',
    y='bedrooms',
    trendline='ols',
    title='Relationship between rental price and the number of rooms'
)

fig.show()
```

```python
pearson_corr2 = relationship['price'].corr(
    relationship['bedrooms'],
    method='pearson'
)
print(f"Pearson correlation: {pearson_corr2}")
```

```python
spearman_corr2 = relationship['price'].corr(
    relationship['bedrooms'],
    method='spearman'
)
print(f"Spearman correlation: {spearman_corr2}")
```

#### The hypothesis that the relationship between rent and the number of rooms is inversely proportional is rejected; Pearson’s (0.471) and Spearman’s (0.474) correlation coefficients indicate a positive relationship, which is weak in linear terms and moderate in monotonic terms (Spearman).
