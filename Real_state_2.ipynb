# 1. Import libraries

```python
# System command library
import os 
# Data handling libraries
import pandas as pd 
import numpy as np 
# Machine learning libraries.
# To split the training set into training and test sets.
from sklearn.model_selection import train_test_split
# To build a model using the linear regression algorithm
from sklearn.linear_model import LinearRegression
# To evaluate model performance using MSE, MAE, and R² metrics
from sklearn.metrics import mean_squared_error, mean_absolute_error, r2_score
# To generate a statistical report that can be used to determine the importance of explanatory variables.
import statsmodels.api as sm 
```

# 2. Load data

```python
data=pd.read_csv('venta_inmuebles_data.csv',sep=',')
```

```python
data.head()
```

# 3. Data preparation

```python
data_t = data
```

```python
data_t[data_t.isnull().any(axis=1)]
```

There are no missing values.

```python
data_t[data_t.duplicated()]
```

```python
data_t=data_t.drop_duplicates()
```

Remove duplicates

Prior to model construction, a data quality check was performed. No missing values were found in the dataset, and duplicate rows were identified and removed. Additionally, all variables are numeric, which meets the necessary requirements for implementing a linear regression model. Therefore, the data is in optimal condition to proceed with the modeling phase.

# 4. Model Building

```python
# The target variable is selected, in this case "precio".
Y=data_t['precio']
# The “precio” variable is removed from the dataset.
X=data_t.drop(['precio'], axis=1)
```

```python
X.head()
```

```python
Y.head()
```

Split the data into a training set and a test set. This division allows you to determine the model's generalization performance on new data.

Training data: used to train the model with the learning algorithm.
Test data: used to evaluate the model's fit.

```python
# 20% of the data is set aside for testing.
X_train, X_test, Y_train, Y_test = train_test_split( X, Y, test_size=0.2, random_state=0)
```

```python
# Create the object to build the model
modelo_regresion = LinearRegression()
modelo_regresion
```

```python
# Train the model using the training set.
modelo_regresion.fit(X_train,Y_train)
```

# 5. Model evaluation

```python
# First, I retrieve the metrics from the training data
y_pred = modelo_regresion.predict(X_train)

print(f"MSE: {mean_squared_error(Y_train, y_pred):.2f}")
print(f"MAE: {mean_absolute_error(Y_train, y_pred):.2f}")
print(f"R²: {r2_score(Y_train, y_pred):.2f}")
```

```python
y_pred = modelo_regresion.predict(X_test)

print(f"MSE: {mean_squared_error(Y_test, y_pred):.2f}")
print(f"MAE: {mean_absolute_error(Y_test, y_pred):.2f}")
print(f"R²: {r2_score(Y_test, y_pred):.2f}")
```

# 7. Conclusions

#### Overall Model Performance
The model explains 90% of the price variability (R-squared = 0.900), indicating excellent predictive power.
The model is statistically significant, with an F-statistic of 5470 and a p-value of 0.00. This confirms that the independent variables collectively have explanatory power over the price.

#### Variable Significance
Most variables are statistically significant, with the following exceptions:

- `numero_baños` (p = 0.715): Not significant. Consider removing this variable.
- `area_del_lote_pie2` (p = 0.057): Very close to the 0.05 threshold. Could be considered marginally significant.

#### Negative Coefficients
Two variables show unexpected negative coefficients:

| Variable | Coefficient | Interpretation |
|----------|-------------|----------------|
| `numero_cuartos` | -19,030 € | Each additional room decreases the price |
| `año_de_construcción` | -179 € | Newer properties decrease in price |

These counterintuitive results are likely explained by multicollinearity (see below).

#### Model Diagnostics

| Test | Value | Interpretation |
|------|-------|----------------|
| **Durbin-Watson** | 1.99 | Close to 2 → No evidence of strong autocorrelation in residuals. Good. |
| **Omnibus (p)** | 0.00 | Residuals do not follow a perfect normal distribution |
| **Jarque-Bera (p)** | 0.00 | Confirms non-normality of residuals |
| **Condition Number** | 148,000 | Very high → Indicates severe multicollinearity |

#### Multicollinearity Issue
The high condition number (148,000) suggests:

- Predictor variables are correlated with each other
- Large scale differences between variables
- Potential numerical instability
- Explains the negative coefficients observed in `numero_cuartos` and `año_de_construcción`

#### Recommendations

1. Check multicollinearity using VIF (Variance Inflation Factor)
2. Consider removing `numero_baños` (not significant)
3. Standardize variables to address scale differences
4. Eliminate redundant variables to reduce multicollinearity
5. Consider transformations (e.g., log of price) to address non-normal residuals
