We need to change the data type of the ‘fecha_envio_limite’ variable to datetime

```python
data_csv['fecha_envio_limite'] = pd.to_datetime(data_csv['fecha_envio_limite'].astype(str), dayfirst=True)
data_csv['fecha_envio_limite']
```

### Days with the highest sales

```python
data_csv['fecha_envio_limite'] = pd.to_datetime(data_csv['fecha_envio_limite'], dayfirst=True)
data_csv['day_month'] = data_csv['fecha_envio_limite'].dt.strftime('%d-%m')

shopping_per_day = data_csv['day_month'].value_counts()

print(shopping_per_day.head())
```

# xslx file

```python
data_excel_1 = pd.read_excel('productos_por_fecha.xlsx', sheet_name='Datos 1')
data_excel_1.head()
```

```python
data_excel_2 = pd.read_excel('productos_por_fecha.xlsx', sheet_name='Datos 2')
data_excel_2.head()
```

In this case, two datasets from the same Excel file were loaded; they were on different sheets. As you can see, some of the column names in “data_excel_1” are incorrectly spelled.

```python
print(data_excel_1.shape)
print(data_excel_2.shape)
```

```python
print(data_excel_1.dtypes)
```

```python
print(data_excel_2.dtypes)
```

We converted the format of the ‘fecha_envio_limite’ column to the datetime type, again

```python
data_excel_1['fecha_envio_limite'] = pd.to_datetime(data_excel_1['fecha_envio_limite'], dayfirst=True)
data_excel_2['fecha_envio_limite'] = pd.to_datetime(data_excel_2['fecha_envio_limite'], dayfirst=True)
```

### Days with the highest sales

```python
for i, df in enumerate([data_excel_1, data_excel_2], 1):
    days = pd.to_datetime(df['fecha_envio_limite'], dayfirst=True).dt.strftime('%d-%m').value_counts()
    print(f"Data {i} - Best day: {days.index[0]} ({days.iloc[0]} ventas)")
    print(days.head(), "\n")
```

### Best-selling categories

```python
for i, df in enumerate([data_excel_1, data_excel_2], 1):
    category = df['nombre_categoria_producto'].value_counts()
    print(f"Data {i} - Top category: '{category.index[0]}' ({category.iloc[0]} times)")
```

# json file

```python
f = open('productos_por_fecha.json')

data_json = json.load(f)

show_data = {}
for i in data_json.keys():
    show_data[i] = dict(list(data_json[i].items()))

show_data
```
