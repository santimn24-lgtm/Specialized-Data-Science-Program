# Introduction

The goal of this notebook is to combine and integrate data from multiple sources using join and merge operations in pandas.

The dataset `empleadosTecnologiaAlpesMod3y4.xlsx` contains several sheets. The main sheet, “employeesTechnologyAlps,” stores information about employees, including the numerical codes for the department (`ID Departamento`) and the position (`Rol ID`). The “Departamentos” and “Roles” sheets contain the corresponding descriptive labels. 

The “Viajes por Empleado” sheet contains the pre-organized list of employee travel frequency, so the join function will be used to append this information to the final dataset.

This approach provides hands-on experience in combining datasets, handling mismatched keys, and understanding how different types of joins affect the final result.

# Import libraries

```python
import pandas as pd
```

# merge exercise

```python
employee_data=pd.read_excel('empleadosTecnologiaAlpesMod3y4.xlsx', sheet_name='empleadosTecnologíaAlpes')
employee_data.head(5)
```

```python
employee_data.dtypes
```

```python
employee_data = employee_data[['Edad', 'Retirado', 'ID Departamento', 'Rol ID']]
employee_data.head(5)
```

The primary keys are 'ID Departamento' and 'Rol ID'

```python
data_Roles=pd.read_excel('empleadosTecnologiaAlpesMod3y4.xlsx', sheet_name='Roles')
data_Roles.head(5)
```

```python
data_Departments=pd.read_excel('empleadosTecnologiaAlpesMod3y4.xlsx', sheet_name='Departamentos')
data_Departments.head(5)
```

```python
data_final = employee_data.merge(
    data_Departments,
    left_on='ID Departamento',
    right_on='ID',
    how='left')
data_final.head()
```

```python
data_final = data_final.merge(
    data_Roles,
    left_on='Rol ID',
    right_on='ID',
    how='left')
data_final.head()
```

```python
data_final = data_final.drop(columns=['ID_x', 'ID Departamento','ID_y','Rol ID'])
data_final.head()
```

Let's explore the dataset

```python
data_final['Departamento'].value_counts()
```

```python
data_final['Rol'].value_counts()
```

```python
data_final.groupby('Rol')['Departamento'].unique()
```

```python
data_final.pivot_table(index='Rol', columns='Departamento', aggfunc='size', fill_value=0)
```

# join exercise

```python
data_Travel=pd.read_excel('empleadosTecnologiaAlpesMod3y4.xlsx', sheet_name='Viajes por Empleado')
data_Travel.head(5)
```

```python
data_final = data_final.reset_index(drop=True)
data_Travel = data_Travel.reset_index(drop=True)

data_final = data_final.join(data_Travel)
data_final.head()
```

```python
data_final.pivot_table(index='Rol', columns='ViajesDeNegocio', aggfunc='size', fill_value=0)
```
