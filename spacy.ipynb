```python
import spacy 
import pandas as pd
```

```python
# We loaded the Spanish NLP model from Spacy
nlp = spacy.load("es_core_news_md")
```

```python
dataframe = pd.read_csv("hotelsb_df_es.csv", sep=',', encoding='utf-16')
dataframe.head()
```

# 3. spaCy

#### Text tokenization

```python
# We create a document using the first record
doc = nlp(dataframe.iloc[0].description) #Oracion formada con el diccionario de palabras NLP
for token in doc:
    print(token.text)
```

```python
# Let's save our processed documents in a list

nlp_docs = []

for _, hotel in dataframe.iterrows():
    doc = nlp(hotel.description)
    nlp_docs.append(doc)
    
# Show the fourth document (index 3) in the list
nlp_docs[3]
```

#### parse

```python
for token in nlp_docs[3]: #Iteramos sobre cada token y extraemos los siguientes atributos de un documento
    print(token.text, token.lemma_, token.pos_, token.tag_, token.dep_,
            token.shape_, token.is_alpha, token.is_stop)
```

#### Visualization of how sentences are interpreted

```python
from spacy import displacy 
sentence_spans = list(nlp_docs[3].sents) 
```

```python
displacy.render(sentence_spans, style='dep', jupyter=True, options={'distance': 140})
```

```python
displacy.render(sentence_spans, style='ent', jupyter=True)
```

#### Comparison between descriptions

Description row 1 versus row 2

```python
print(nlp_docs[1], "<->", nlp_docs[2], nlp_docs[1].similarity(nlp_docs[2]))
```

Description row 2 versus row 3

```python
print(nlp_docs[2], "<->", nlp_docs[3], nlp_docs[2].similarity(nlp_docs[3]))
```
